# Construção dos indicadores



```
variavel: tx_internacoes_1000hab
descricao: Número de internações hospitalares por mil habitantes
formula: (Internacoes / populacao) * 1000
unidade: Internações por 1.000 habitantes
finalidade: Mensurar a utilização hospitalar relativa ao tamanho da população
    
variavel: gasto_hospitalar_per_capita
descricao: Valor total das AIHs aprovadas por habitante
formula: Valor_Total / populacao
unidade: R$ por habitante
finalidade: Mensurar o gasto hospitalar relativo ao tamanho da população
    
variavel: valor_medio_internacao
descricao: Valor médio associado a cada internação hospitalar
formula: Valor_Total / Internacoes
unidade: R$ por internação
finalidade: Mensurar o custo médio das internações

variavel: tx_mortalidade_hospitalar
descricao: Proporção de internações associadas a óbito
formula: (Obitos / Internacoes) * 100
unidade: %
finalidade: Mensurar a mortalidade hospitalar

variavel: razao_cadunico_pop
descricao: Razão entre o número de pessoas registradas no CadÚnico e a população municipal utilizada como referência.
formula: (pessoas_cadunico / populacao) * 100
unidade: %
finalidade: Indicador de vulnerabilidade socioeconômica municipal

variavel: razao_aih_internacao
descricao: Razão entre AIHs aprovadas e internações hospitalares
formula: AIH_Aprovadas / Internacoes
unidade: AIH por internação
finalidade: Indicador exploratório da relação entre AIHs processadas e internações
```



````python
# Internações por 1.000 habitantes
df['internacoes_1000hab'] = (
    df['Internacoes'] /
    df['populacao'] * 1000
)
````

````python
# Valor hospitalar per capita
df['valor_per_capita'] = (
    df['Valor_Total'] /
    df['populacao']
)
````

````python
# Valor médio por internação
df['valor_medio_internacao'] = np.where(
    df['Internacoes'] > 0,
    df['Valor_Total'] /
    df['Internacoes'],
    np.nan
)
````

````python
# Óbitos por 1.000 internações
df['obitos_1000internacoes'] = np.where(
    df['Internacoes'] > 0,
    df['Obitos'] /
    df['Internacoes'] * 1000,
    np.nan
)
````

````python
# Razão entre o número de pessoas registradas no CadÚnico e a população
df['razao_cadunico_pop'] = (
    df['pessoas_cadunico'] /
    df['populacao'] * 100
)
````

````python
# Relação AIHs/internações
df['aih_por_internacao'] = np.where(
    df['Internacoes'] > 0,
    df['AIH_Aprovadas'] /
    df['Internacoes'],
    np.nan
)
````

````python
# AIHs por 1.000 habitantes
df['aih_1000hab'] = (
    df['AIH_Aprovadas'] /
    df['populacao'] * 1000
)
````

````python
ipca = pd.read_csv('/content/drive/MyDrive/TCC/TCC/dados_brutos/IBGE/df_ipca_2017_2025.csv')
````

````python
ipca.head()
````

````python
df_final = df.merge(
    ipca[['ano', 'mes', 'ipca_mes_anterior']],
    on=['ano', 'mes'],
    how='left'
)
````

````python
df_final.info()
````

````python
df_final['ipca_mes_anterior'].isna().sum()
````

````python
ipca_dez2025 = ipca.loc[
    (ipca['ano'] == 2025) &
    (ipca['mes'] == 12),
    'ipca'
].iloc[0]
````

````python
ipca_dez2025
````

````python
df_final[
    ['ano', 'mes', 'ipca_mes_anterior']
].drop_duplicates().sort_values(['ano', 'mes']).head()
````

````python
df_final['fator_ipca'] = (
    ipca_dez2025 /
    df_final['ipca_mes_anterior']
)
````

````python
df_final[
    ['ano', 'mes', 'ipca_mes_anterior', 'fator_ipca']
].drop_duplicates().sort_values(['ano', 'mes'])
````

````python
df_final['Valor_Total_real'] = (
    df_final['Valor_Total'] *
    df_final['fator_ipca']
)
````

````python
df_final['valor_medio_internacao_real'] = np.where(
    df_final['Internacoes'] > 0,
    df_final['Valor_Total_real'] /
    df_final['Internacoes'],
    np.nan
)
````

````python
df_final['valor_per_capita_real'] = (
    df_final['Valor_Total_real'] /
    df_final['populacao']
)
````

````python
df_final.to_csv(
    '/content/drive/MyDrive/TCC/TCC/base_processada.csv',
    index=False,
    encoding='utf-8-sig'
)
````

````python
# Quantidade e percentual de registros com zero óbitos
n_zeros = (df_final['Obitos'] == 0).sum()

print(f"Registros com zero óbitos: {n_zeros:,}")
print(f"Percentual: {n_zeros / len(df_final) * 100:.2f}%")

# Distribuição por ano
df_final.groupby('ano')['Obitos'].agg(
    total='count',
    registros_zero=lambda x: (x == 0).sum(),
    registros_positivos=lambda x: (x > 0).sum(),
    soma_obitos='sum'
)
````

````python
df_obitos_anual = (
    df_final
    .groupby(['codigo_municipio', 'municipio', 'ano'], as_index=False)
    .agg(
        obitos=('Obitos', 'sum'),
        internacoes=('Internacoes', 'sum')
    )
)

df_obitos_anual['obitos_1000internacoes'] = (
    df_obitos_anual['obitos'] /
    df_obitos_anual['internacoes'].replace(0, np.nan) * 1000
)

df_obitos_anual.info()
````

````python
print("Total de municípios-ano:", len(df_obitos_anual))

print(
    "Municípios-ano com zero óbitos:",
    (df_obitos_anual['obitos'] == 0).sum()
)

print(
    "Municípios-ano com zero internações:",
    (df_obitos_anual['internacoes'] == 0).sum()
)

print(
    "Taxas anuais nulas:",
    df_obitos_anual['obitos_1000internacoes'].isna().sum()
)

# Resumo por ano
df_obitos_anual.groupby('ano').agg(
    municipios=('codigo_municipio', 'nunique'),
    municipios_zero_obitos=('obitos', lambda x: (x == 0).sum()),
    total_obitos=('obitos', 'sum'),
    total_internacoes=('internacoes', 'sum')
)
````
