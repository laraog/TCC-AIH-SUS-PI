```python
anos_padrao = {
    2018: ('POP2018_20220905.xls', 'Municípios'),
    2019: ('POP2019_20220905.xls', 'Municípios'),
    2020: ('POP2020_20220905.xls', 'Municípios'),
    2021: ('POP2021_20240624.xls', 'Municípios'),
    2023: ('POP_DOU_2023_Municipios_POP2022_Malha2023.xls', 'Municípios'),
    2024: ('POP2024_20241230.xls', 'MUNICÍPIOS'),
    2025: ('POP2025_20260828.xls', 'Municípios')
}

lista_pop = []
```

```python
for ano, (nome_arquivo, aba) in anos_padrao.items():

    caminho = os.path.join(pasta, nome_arquivo)

    df = pd.read_excel(
        caminho,
        sheet_name=aba,
        header=1,
        dtype={
            'UF': str,
            'COD. UF': str,
            'COD. MUNIC': str
        }
    )

    # mantém apenas Piauí
    df = df[df['UF'] == 'PI'].copy()

    # remove colunas totalmente vazias
    df = df.dropna(axis=1, how='all')

    # padroniza nomes
    df = df.rename(columns={
        'NOME DO MUNICÍPIO': 'municipio',
        'POPULAÇÃO ESTIMADA': 'populacao',
        'POPULAÇÃO': 'populacao'
    })

    # garante tamanhos corretos dos códigos
    df['COD. UF'] = df['COD. UF'].str.zfill(2)
    df['COD. MUNIC'] = df['COD. MUNIC'].str.zfill(5)

    # cria código IBGE completo
    df['codigo_ibge'] = df['COD. UF'] + df['COD. MUNIC']

    # adiciona ano
    df['ano'] = ano

    # mantém só o necessário
    df = df[
        ['codigo_ibge', 'municipio', 'ano', 'populacao']
    ]

    lista_pop.append(df)
```

```python
caminho_2022 = os.path.join(
    pasta,
    'populacao_censo_2022.xlsx'
)

df_2022 = pd.read_excel(
    caminho_2022,
    sheet_name='Tabela',
    header=None
)

# pega apenas os dados a partir da linha 4
df_2022 = df_2022.iloc[4:, :3].copy()

df_2022.columns = [
    'codigo_ibge',
    'municipio',
    'populacao'
]

# tipos
df_2022['codigo_ibge'] = (
    df_2022['codigo_ibge']
    .astype(str)
    .str.replace(r'\.0$', '', regex=True)
    .str.zfill(7)
)

df_2022['populacao'] = pd.to_numeric(
    df_2022['populacao'],
    errors='coerce'
)

# remove "(PI)" do município
df_2022['municipio'] = (
    df_2022['municipio']
    .str.replace(r'\s*\(PI\)$', '', regex=True)
)

df_2022['ano'] = 2022

df_2022 = df_2022[
    ['codigo_ibge', 'municipio', 'ano', 'populacao']
]
```

```python
#Função para limpar as tabelas do DATASUS

def limpar_datasus(caminho_arquivo, nome_da_coluna_valor):
    df = pd.read_csv(caminho_arquivo, sep=';', encoding='latin1', skiprows=3)

    if 'Total' in df.columns:
        df = df.drop(columns=['Total'])

    df_long = pd.melt(df, id_vars=['Município'], var_name='Mes_Ano', value_name=nome_da_coluna_valor)

    df_long = df_long[df_long['Município'].str.startswith('22', na=False)].copy()

    df_long['Cod_Municipio'] = df_long['Município'].str[:6]
    df_long['Nome_Municipio'] = df_long['Município'].str[7:]

    df_long = df_long.drop(columns=['Município'])

    return df_long

print("Função 'limpar_datasus' criada com sucesso!")
```

```python
nomes_colunas = {
    'sih_aih_aprovadas_18_25_atend.csv': 'AIH_Aprovadas',
    'sih_internacoes_18_25_atend.csv': 'Internacoes',
    'sih_valor_total_18_25_atend.csv': 'Valor_Total',
    'sih_dias_perm_18_25_atend.csv': 'Dias_Permanencia',
    'sih_media_perm_18_25_atend.csv': 'Media_Permanencia',
    'sih_obitos_18_25_atend.csv': 'Obitos'
}
```

```python
dados_limpos = {}

for arquivo, nome_coluna in nomes_colunas.items():
    caminho = os.path.join(pasta, arquivo)

    dados_limpos[nome_coluna] = limpar_datasus(
        caminho,
        nome_coluna
    )
```
